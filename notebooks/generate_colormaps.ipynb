```python
# main path
data_folder = '../data'

# paths
cmaps_path = join(data_folder, 'colormaps')
```

# Main code

## Generate colormaps

```python
cmap_blues = mcolors.LinearSegmentedColormap.from_list('calif_blues', ['#ffffff', '#1E4D8B'], N=255)
cmap_oranges = mcolors.LinearSegmentedColormap.from_list('calif_yellows', ['#ffffff', '#B8860B'], N=255)
cmap_orange_blue = mcolors.LinearSegmentedColormap.from_list('calif_ywb', ['#B8860B', '#ffffff', '#1E4D8B'], N=255)

for cmap in (cmap_blues, cmap_oranges, cmap_orange_blue):
    # Generate 255 colors from the colormap
    color_array = cmap(np.linspace(0, 1, 255))

    # Convert to uint8 format (0-255) and ensure RGBA structure
    cmap_data = (color_array * 255).astype('uint8')

    # Save as Terracotta-compatible colormap
    np.save(join(cmaps_path, f'{cmap.name}_rgba.npy'), cmap_data)
```

## Generate thumbnail

```python
for cmap in (cmap_blues, cmap_oranges, cmap_orange_blue):
    # Create a figure with the exact size we need
    # DPI=100, so figsize in inches = pixels/100
    fig, ax = plt.subplots(figsize=(2.56, 0.24), dpi=100)
    
    # Remove all margins and padding
    fig.subplots_adjust(left=0, right=1, top=1, bottom=0)
    
    # Create a gradient from left to right
    gradient = np.linspace(0, 1, 256).reshape(1, -1)
    
    # Display the gradient
    ax.imshow(gradient, aspect='auto', cmap=cmap)
    
    # Remove axes
    ax.set_axis_off()
    
    # Save as PNG
    plt.savefig(join(cmaps_path, f'cmap-{cmap.name}-bar.webp'), dpi=100, bbox_inches='tight', pad_inches=0)
    plt.close()
```
